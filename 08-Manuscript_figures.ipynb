```python
#metadata

md = pd.read_csv("/mnt/disks/data/imc/CART_cohort/in/sample_md.csv")
md['cohort_batch'] = md['cohort'].astype(str)+'_'+md['batch'].astype(str).fillna('').astype('str')
md['cohort_batch'] = md['cohort_batch'].map(
    {'original_cohort_nan':"original_cohort", 'original_cohort_second_acquisition_nan':"original_cohort_second_acquisition",
       'validation_cohort_batch1':"validation_cohort_batch1", 'validation_cohort_batch2':"validation_cohort_batch2"}
)
md['cohort_marker_indexing'] = md['cohort_batch'].map(
    {'original_cohort':'original_cohort','original_cohort_second_acquisition':'validation_cohort',
     'validation_cohort_batch1':'validation_cohort','validation_cohort_batch2':'validation_cohort'}
)
panel_og = pd.read_csv("/mnt/disks/data/imc/CART_cohort/in/panel.csv")
panel_valid = pd.read_csv("/mnt/disks/data/imc/CART_cohort/in/panel_valid.csv")
shared_markers = list( set(panel_og['name'].tolist()) & set(panel_valid['name'].tolist()) ) 
panels = {'original_cohort':panel_og, 'validation_cohort':panel_valid}

#funs: import/prep

def load_img(sample_id):
    cohort_marker_indexing = md[md['sample']==sample_id]['cohort_marker_indexing'].tolist()[0]
    img = tifffile.imread(f'/mnt/disks/data/imc/CART_cohort/out/sample_data/{sample_id}/corrected.tif')
    #re-index img
    protein_indices = [ panels[cohort_marker_indexing]['name'].tolist().index(p) for p in shared_markers ]
    img = img[protein_indices]
    return img

def load_mask(sample_id):
    return tifffile.imread(f'/mnt/disks/data/imc/CART_cohort/out/sample_data/{sample_id}/segmentation_masks/mask_V1withAdipo.tif')

def load_gpd(sample_id):
    return gpd.read_file(f'/mnt/disks/data/imc/CART_cohort/out/sample_data/{sample_id}/segmentation_masks/mask_V1withAdipo_gdf.geojson')

def prep_gdf(sample_id):
    #import md
    obs_v5 = pd.read_csv('/mnt/disks/data/imc/CART_cohort/out/v1_pheno_work/obs_v5.csv')
    obs_sub = obs_v5[obs_v5.sample_id==sample_id][['Cell_ID','lineage','ph','gated','Ki67is','PDL1is']]
    #gdf,mask
    gdf = load_gpd(sample_id)
    mask = tifffile.imread(f'/mnt/disks/data/imc/CART_cohort/out/sample_data/{sample_id}/segmentation_masks/mask_V1withAdipo_MKv2.tif')
    mask_MK = np.where(np.isin(mask,obs_sub[obs_sub.lineage=='MKs'].Cell_ID),mask,0)
    #merge
    gdf = gdf.merge(obs_sub, how='left', left_on='id',right_on='Cell_ID')
    gdf.ph = gdf.ph.fillna("Artifact")
    #re-calc for MKs
    gdf_nonMK = gdf[~((gdf.ph.isna())|(gdf.ph=='MK'))]
    gdf_MK = mask2shapely(mask_MK)
    gdf_MK = gdf_MK.drop(['x','y'],axis=1).merge(obs_sub, how='left', left_on='id',right_on='Cell_ID')
    #merge final
    gdf = pd.concat([gdf_nonMK,gdf_MK])
    return gdf

def load_data(sample_id):
    '''img,markers,mask,gdf = load_data(sample_id)'''
    return load_img(sample_id), shared_markers, load_mask(sample_id), prep_gdf(sample_id)

def mask2shapely(masks_in):
    from shapely.geometry import Polygon
    import geopandas as gpd
    from skimage import measure
    import numpy as np
    
    ids = []
    geometries = []

    for i in np.unique(masks_in)[1:]:
        binary_mask = (masks_in == i).astype(np.uint8)
        
        # Find contours using scikit-image
        contours = measure.find_contours(binary_mask, 0.5)
        
        # If multiple contours (disconnected pixels), take the longest one
        if len(contours) > 1:
            longest_contour = max(contours, key=len)
            # Convert from (row, col) to (x, y) coordinates
            coords = longest_contour[:, [1, 0]]
            if len(coords) > 3:
                geometries.append(Polygon(coords))
                ids.append(i)
        else:
            if len(contours) > 0 and len(contours[0]) > 3:
                # Convert from (row, col) to (x, y) coordinates
                coords = contours[0][:, [1, 0]]
                geometries.append(Polygon(coords))
                ids.append(i)
    
    gdf = gpd.GeoDataFrame({'id': ids, 'geometry': geometries}, crs='EPSG:4326')
    gdf = gdf[gdf['id'] != 0]
    gdf['x'] = gdf.geometry.centroid.x
    gdf['y'] = gdf.geometry.centroid.y   

    return gdf

#funs: plotting 

def img_map(img,figsize=3,cmap='viridis',title=None, origin='lower',labels=False,cbar=False,save=None):
    fig, ax = plt.subplots(figsize=(figsize,figsize))
    im = ax.imshow(img, cmap=cmap, origin=origin)
    ax.set_title(title)
    if not labels:
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_xticklabels([])
        ax.set_yticklabels([])
    if cbar:
        plt.colorbar(im, ax=ax)
    if save is not None:
        plt.savefig(save, dpi=300, bbox_inches='tight', transparent=False) 
    plt.show()
    
def pctnorm(img, pct_lo = 1, pct_hi = 99.7):
    from skimage import exposure
    p_low, p_high = np.percentile(img, [pct_lo, pct_hi])
    return exposure.rescale_intensity(img, in_range=(p_low, p_high), out_range=(0, 1))

def pctnorm_stack(stack, pct_lo = 1, pct_hi = 99.7):
    img_norm = np.zeros_like(stack, dtype=np.float32)
    for i in range(stack.shape[0]):
        img_norm[i] = pctnorm(stack[i], pct_lo=pct_lo, pct_hi=pct_hi)
    return img_norm

def mi(m):
    if isinstance(m, str):
        return markers.index(m)
    elif isinstance(m, list):
        return [markers.index(i) for i in m]


def mask_map(img,mask, figsize=(4,4),fs=None, linewidth=0.75, title=None, origin='upper', cmap='viridis', labels=False):
    if fs is not None:
        figsize=fs


    gdf = mask2shapely(mask)
    
    fig,ax = plt.subplots(figsize=figsize)
    ax.imshow(img, cmap=cmap, interpolation='none', origin=origin)
    gdf.boundary.plot(aspect=1, ax=ax, color='white',linewidth=linewidth)
    ax.set_title(title)
    if not labels:
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_xticklabels([])
        ax.set_yticklabels([])
        
def plot_multichannel_image(channels,ax, global_ranges=None, gamma=1.0, 
                           channel_names=None, channel_colors=None, linewidth=0.25):

    n_channels = len(channels)
    if n_channels < 2 or n_channels > 4:
        raise ValueError("Need 2-4 channels")
    
    height, width = channels[0].shape
    
    # Map channels to RGB (supports 2-4 channels)
    rgb_map = {
        2: [2, 1, None],      # R, G
        3: [2, 1, 0],          # R, G, B
        4: [2, 1, 0]           # R, G, B (4th channel ignored or merged)
    }
    
    rgb_image = np.zeros((height, width, 3), dtype=np.uint8)
    
    for i, (ch, rgb_idx) in enumerate(zip(channels[:3], rgb_map[n_channels])):
        if rgb_idx is None:
            continue
            
        # Determine scaling range
        if global_ranges is not None:
            vmin, vmax = global_ranges[i]
        else:
            vmin, vmax = ch.min(), ch.max()
        
        # Scale to 0-255
        clipped = np.clip(ch, vmin, vmax)
        scaled = ((clipped - vmin) / (vmax - vmin) * 255)
        
        # Apply gamma if needed
        if gamma != 1.0:
            scaled = ((scaled / 255) ** gamma) * 255
            
        rgb_image[:, :, rgb_idx] = scaled.astype(np.uint8)
    
    # Display image
    ax.imshow(rgb_image)
    ax.set_xticks([])
    ax.set_yticks([])

    # Display  mask
    ####gdf.boundary.plot(aspect=1, ax=ax, color='white',linewidth=linewidth)
    
    # Add legend below plot if names provided
    if channel_names:
        if channel_colors is None:
            channel_colors = ['red', 'green', 'blue', 'cyan'][:n_channels]
        
        # Add black background box below axis
        ax.text(0.5, -0.08, '', 
               transform=ax.transAxes,
               ha='center', va='center',
               bbox=dict(boxstyle='round,pad=0.4', 
                        facecolor='black', 
                        edgecolor='none',
                        alpha=0.9),
               zorder=10)
        
        # Add individual colored labels more tightly spaced
        n_labels = len(channel_names)
        total_width = 0.6  # Use 60% of axis width for labels
        if n_labels > 1:
            spacing = total_width / (n_labels - 1)
            start_pos = 0.5 - total_width / 2
        else:
            spacing = 0
            start_pos = 0.5
            
        for i, (name, color) in enumerate(zip(channel_names, channel_colors)):
            xpos = start_pos + spacing * i
            ax.text(xpos, -0.08, name,
                   transform=ax.transAxes,
                   ha='center', va='center',
                   color=color, fontsize=10, weight='bold',
                   zorder=11)
    
    return ax
    
def plot_gdf_pheno(img, gdf, ax, ncol=1):

    import matplotlib.patches as mpatches

    ax.imshow(img, cmap='gist_grey')
    
    gdf.plot(aspect=1, ax=ax, color=gdf["color"], edgecolor='black', linewidth=0.5, alpha=1)
    
    unique_entries = gdf.drop_duplicates(subset=["label", "color"])
    handles = [
        mpatches.Patch(facecolor=row["color"], edgecolor='black', label=row["label"])
        for _, row in unique_entries.iterrows()
    ]
    ax.legend(handles=handles,bbox_to_anchor=(1.05, 1), ncol=ncol,loc="upper left")

    
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    ax.set_axis_off()
    
def make_marker_plot(img,mask,ms,xmin,ymin,dim,label,labels_pl,label_colors,
                     save=None,
                     clus_exclude=['B-T', 'Artifact']):

    img_norm = pctnorm_stack(img[mi(ms)])[:,ymin:ymin+dim,xmin:xmin+dim]
    img_dummy = np.zeros(img_norm[0].shape)

    #new gdf
    gdf_sub = mask2shapely(mask[ymin:ymin+dim,xmin:xmin+dim])
    gdf_sub  = gdf_sub.merge(gdf[['id','gated']+[label]])

    #prep gdf for plotting
    gdf_sub['label'] = np.where(np.isin(gdf_sub[label],labels_pl),gdf_sub[label],'Other')
    gdf_sub['color']=gdf_sub['label'].map(label_colors)
    
    #standard QC
    gdf_sub = gdf_sub[~gdf_sub.gated.isna()]
    gdf_sub = gdf_sub[gdf_sub.gated]
    gdf_sub = gdf_sub[~gdf_sub[label].isin(clus_exclude)]

    #Make plot
    fig,axes = plt.subplots(figsize=(10,5),ncols=2,constrained_layout=True)
    plot_multichannel_image(img_norm, axes[0], channel_names=ms,channel_colors=['blue','#00cc00','red'])
    plot_gdf_pheno(img_dummy, gdf_sub, axes[1])
    plt.tight_layout()
    plt.savefig(save+'.pdf', dpi=300, bbox_inches='tight', transparent=False)
    plt.savefig(save+'.svg', dpi=300, bbox_inches='tight', transparent=False)
    plt.show()

def ph_cn_multiplot(mask_in, CN_, save=None,  linewidth=1.5):
    
    gdf_sub=mask2shapely(mask_in)
    gdf_sub  = gdf_sub.merge(gdf[['id','gated','ph']])
    gdf_sub = gdf_sub.merge(cn_cells_data.rename({'Cell_ID':'id'},axis=1)[['id','cn_labels']], how='left')
    gdf_sub.cn_labels = gdf_sub.cn_labels.fillna('NA')
    
    fig,axes = plt.subplots(figsize=(8,4),ncols=2)
    
    gdf_sub['label'] = np.where(gdf_sub['cn_labels'].isin(CN_),gdf_sub['cn_labels'],'NA')
    gdf_sub['color']=gdf_sub['label'].map(bslCN_col)
    gdf_sub.plot(aspect=1, ax=axes[0], color=gdf_sub['color'], alpha=1, edgecolor='black', linewidth=linewidth)
    
    gdf_sub['label'] = gdf_sub['ph'].map({**cell_type_renamer,**{'B-T':'Unclassified','Artifact':'Unclassified'}})
    gdf_sub['color']=gdf_sub['label'].map(ph_colors)
    gdf_sub.plot(aspect=1, ax=axes[1], color=gdf_sub['color'], alpha=1, edgecolor='black', linewidth=linewidth)
    
    for ax in axes:
        bounds = gdf_sub.total_bounds  # [minx, miny, maxx, maxy]
        pad=0
        ax.set_xlim(bounds[0]-pad, bounds[2]+pad)
        ax.set_ylim(bounds[1]-pad, bounds[3]+pad)
        ax.set_xticks([])
        ax.set_yticks([])
        
    plt.tight_layout()
    if save is not None:
        plt.savefig(save+'.pdf', dpi=300, bbox_inches='tight', transparent=False)
        plt.savefig(save+'.svg', dpi=300, bbox_inches='tight', transparent=False)
    plt.show()
```

### Fig.S6C

```python
# CN composition comparison
```

```python
cn_id = "bslTP_linNegOut_15clus"
cn_cells_pre = pd.read_csv(f'/mnt/disks/data/imc/CART_cohort/out/v1_pheno_work/cn_data/{cn_id}-cells_data.csv')
cn_id = "postTP_linNegOut_15clus"
cn_cells_post = pd.read_csv(f'/mnt/disks/data/imc/CART_cohort/out/v1_pheno_work/cn_data/{cn_id}-cells_data.csv')
del cn_id
```

```python
from scipy.stats import pearsonr
from statsmodels.stats.multitest import multipletests

data1=cn_cells_pre[['cn_labels','ph']].value_counts().reset_index()
data2=cn_cells_post[['cn_labels','ph']].value_counts().reset_index()

data = pd.concat([data1, data2], keys=['table1', 'table2']).reset_index(level=0)
data = data.rename(columns={'level_0': 'table'})

# Pivot to wide format (ensure 0s for missing combinations)
pivot = data.pivot_table(
    index='cn_labels',
    columns=['table', 'ph'],
    values='count',
    fill_value=0
).T

# Compute correlation matrix
cns = pivot.columns
n = len(cns)
r2_matrix = np.zeros((n, n))
pval_matrix = np.zeros((n, n))

for i, cn1 in enumerate(cns):
    for j, cn2 in enumerate(cns):
        r, p = pearsonr(pivot[cn1]['table1'], pivot[cn2]['table2'])
        r2_matrix[i, j] = r**2
        pval_matrix[i, j] = p

# FDR correction (exclude diagonal)
mask = ~np.eye(n, dtype=bool)
pvals_flat = pval_matrix[mask]
_, fdr_flat, _, _ = multipletests(pvals_flat, method='fdr_bh')

# Reconstruct FDR matrix
fdr_matrix = np.ones((n, n))
fdr_matrix[mask] = fdr_flat

# Create DataFrames
r2_df = pd.DataFrame(r2_matrix, index=cns, columns=cns)
fdr_df = pd.DataFrame(fdr_matrix, index=cns, columns=cns)

# Create annotations (asterisk for FDR < 0.05)
annot = np.where(fdr_df < 0.05, '*', '')
annot_df = pd.DataFrame(annot, index=cns, columns=cns)
```

```python
fig, ax = plt.subplots(figsize=(6,5.5))

sns.heatmap(r2_df, annot=annot_df, fmt='s', cmap='RdYlBu_r',
            vmin=0, vmax=1, ax=ax, cbar_kws={'label': 'R²'},
            linewidths=0.5, linecolor='gray')

ax.set_title('CN Correlation (R²)\n* FDR < 0.05', fontsize=12)
ax.set_xlabel('postCN', fontsize=12)
ax.set_ylabel('preCN', fontsize=12)

plt.tight_layout()
plt.savefig(loc_+'Fig_S6C.pdf', dpi=300, bbox_inches='tight', transparent=False) 
plt.savefig(loc_+'Fig_S6C.svg', dpi=300, bbox_inches='tight', transparent=False) 
plt.show()
```

### Intra-sample inter-ROI correlation

```python
obs_v5 = pd.read_csv('/mnt/disks/data/imc/CART_cohort/out/v1_pheno_work/obs_v5.csv')
obs_out = obs_v5.copy()
```

```python
clus = 'ph'
clus_remove = ['Artifact','B-T']
sample_remove = ['BS-21-F34416']

df = obs_out
df = df[~df[clus].isin(clus_remove)]
df = df[~df['sample_id'].isin(sample_remove)]
df = df[df[clus]!='Lin-']
df = df[df.gated==True]

df = df[['sample_id',clus]].value_counts().reset_index()\
.merge(df[['sample_id']].value_counts().reset_index().rename({'count':'total'},axis=1))\
.merge( pd.read_csv('/mnt/disks/data/imc/CART_cohort/in/sample_id-clinical.csv') )

df['batch_id'] = df['slideID'].astype(str)+'__'+df['Timepoint'].astype(str)

df['count'] = df['count']/df['total']*100
```

```python
import pandas as pd
from scipy.stats import pearsonr
from itertools import combinations

def correlate_samples_within_batch(df):
    results = []
    
    for batch_id, batch_df in df.groupby('batch_id'):
        samples = batch_df['sample_id'].unique()
        
        # Skip batches with only one sample
        if len(samples) < 2:
            continue
        
        # Get all unique pairs of samples
        for sample1, sample2 in combinations(samples, 2):
            # Pivot to get lineage counts for each sample
            s1_counts = batch_df[batch_df['sample_id'] == sample1].set_index(clus)['count']
            s2_counts = batch_df[batch_df['sample_id'] == sample2].set_index(clus)['count']
            
            # Align by lineage (handle missing lineages with 0)
            lineages = s1_counts.index.union(s2_counts.index)
            s1_aligned = s1_counts.reindex(lineages, fill_value=0)
            s2_aligned = s2_counts.reindex(lineages, fill_value=0)
            
            # Compute correlation
            r, pval = pearsonr(s1_aligned, s2_aligned)
            
            results.append({
                'batch_id': batch_id,
                'sample1': sample1,
                'sample2': sample2,
                'R2': r**2,
                'R': r,
                'pval': pval,
                'n_lineages': len(lineages)
            })
    
    return pd.DataFrame(results)
```

```python
res = correlate_samples_within_batch(df)
```
